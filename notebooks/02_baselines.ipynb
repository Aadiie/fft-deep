```python
from fftdeep.data.arrays import get_arrays

#Xtr stores numerical data of image and ytr stores it's numpy labelled data (0/1) 

Xtr, ytr = get_arrays("train")
Xva, yva = get_arrays("val")
print(Xtr.shape, Xtr.dtype, ytr.shape, ytr.mean())
print(Xva.shape, yva.mean())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score

print("always-REAL accuracy:", (yva == 0).mean())      # expect 0.5066

def mean_rgb(X):
    return X.reshape(len(X), -1, 3).mean(axis=1) / 255.0   # (N, 3)

clf = LogisticRegression(max_iter=1000).fit(mean_rgb(Xtr), ytr)
p = clf.predict_proba(mean_rgb(Xva))[:, 1]
print("mean-RGB val accuracy:", accuracy_score(yva, p > 0.5))
print("mean-RGB val AUC:", roc_auc_score(yva, p))
```
